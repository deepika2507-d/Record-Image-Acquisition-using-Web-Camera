```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import time
from IPython.display import clear_output

# Open webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Webcam not detected")

else:
    print("Webcam connected successfully")

    for i in range(50):

        ret, frame = cap.read()

        if not ret:
            print("Image capture failed")
            break

        # Convert BGR to RGB
        frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

        # Display webcam image
        clear_output(wait=True)
        plt.imshow(frame_rgb)
        plt.axis('off')
        plt.title("Webcam Image Acquisition")
        plt.show()

        time.sleep(0.05)

    # Release webcam
    cap.release()
    print("Webcam released successfully")
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import time
from IPython.display import clear_output

cap = cv2.VideoCapture(0)

for i in range(50):

    ret, frame = cap.read()

    if not ret:
        break

    resized_frame = cv2.resize(frame, (100, 150))

    frame_rgb = cv2.cvtColor(resized_frame, cv2.COLOR_BGR2RGB)

    clear_output(wait=True)

    plt.imshow(frame_rgb)
    plt.axis('off')
    plt.show()

    time.sleep(0.05)

cap.release()
```

```python
import cv2
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Camera not detected!")
else:
    print("Camera started successfully")

    for i in range(50):
        ret, frame = cap.read()

        if not ret:
            print("Cannot read frame")
            break

        rotated_frame = cv2.rotate(
            frame, cv2.ROTATE_90_CLOCKWISE
        )

        frame_rgb = cv2.cvtColor(
            rotated_frame, cv2.COLOR_BGR2RGB
        )

        clear_output(wait=True)
        plt.figure(figsize=(5, 5))
        plt.imshow(frame_rgb)
        plt.title("Rotated Video")
        plt.axis('off')
        plt.show()
        plt.close()

        time.sleep(0.05)

    cap.release()
    print("Camera released")
```
